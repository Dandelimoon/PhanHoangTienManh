W5A28 Viết hàm tính góc ∠ABC (độ) từ tọa độ A(x1,y1), B(x2,y2), C(x3,y3). 

```python
import math
def goc(A, B, C):
    BA=(A[0]-B[0],A[1]-B[1])
    BC=(C[0]-B[0],C[1]-B[1])
    dot= BA[0]*BC[0] + BA[1]*BC[1]
    dai= ((BA[0]**2+BA[1]**2)**0.5) * ((BC[0]**2+BC[1]**2)**0.5)
    cos= abs(dot/dai)
    if -1<= cos <= 1:
        return f"{math.degrees(math.acos(dot/dai)):.2f}"
    else:
        return "False"
A=tuple(map(int, input().split()))
B=tuple(map(int, input().split()))
C=tuple(map(int, input().split()))
goc(A, B, C)
```

```python
def C(n,k):
    if k>n:
        return "False"
    else:
        k_gt=1
        for i in range(2, k+1):
            k_gt*=i
        n_gt=1
        for i in range(2, n+1):
            n_gt*=i
        a_gt=1
        for i in range(2, (n-k)+1):
            a_gt*=i
        return f"{(n_gt)/(k_gt*a_gt):.2f}"
n,k=map(int, input().split())
C(n,k)
```
